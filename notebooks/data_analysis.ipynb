```python
# Create the database connection and cursor
conn = sqlite3.connect('../data/processed/abs_challenge.db')
cursor = conn.cursor()
```

```python
splits_query = """
SELECT
COUNT(*) as total_challenges,
(SELECT COUNT(*) FROM batter_challenge_results WHERE description = 'called_strike') * 1.0 / (SELECT COUNT(*) FROM batter_challenge_results) * 100 as confirmed_pct,
(SELECT COUNT(*) FROM batter_challenge_results WHERE description = 'ball') * 1.0 / (SELECT COUNT(*) FROM batter_challenge_results) * 100 as overturned_pct
FROM batter_challenge_results;
"""



splits_df = pd.read_sql_query(splits_query, conn)
splits_df.head()
```

BATTER CHALLENGE VALUE LEADERS

```python
challenge_value_leaders_query = """
SELECT
player_name,
SUM(challenge_value) as total_challenge_value,
COUNT(*) as total_challenges,
COUNT(CASE WHEN description = 'ball' THEN 1 END) as total_overturned
FROM batter_challenge_results
WHERE challenge_value IS NOT NULL
GROUP BY player_name
ORDER BY total_challenge_value DESC;
"""

challenge_value_leaders_df = pd.read_sql_query(challenge_value_leaders_query, conn)
challenge_value_leaders_df.head(20)
```

LEADERS IN AVERAGE CHALLENGE VALUE PER CHALLENGE (HITTERS)

```python
average_challenge_value_leaders_query = """
SELECT
    player_name,
    AVG(challenge_value) AS average_challenge_value
FROM batter_challenge_results
WHERE challenge_value IS NOT NULL
GROUP BY player_name
HAVING COUNT(*) >= 5
ORDER BY average_challenge_value DESC;
"""

average_challenge_value_leaders_df = pd.read_sql_query(average_challenge_value_leaders_query, conn)
average_challenge_value_leaders_df.head(20)
```

```python
splits_query = """
SELECT
COUNT(*) as total_challenges,
(SELECT COUNT(*) FROM catcher_challenge_results WHERE description = 'ball') * 1.0 / (SELECT COUNT(*) FROM catcher_challenge_results) * 100 as confirmed_pct,
(SELECT COUNT(*) FROM catcher_challenge_results WHERE description = 'called_strike') * 1.0 / (SELECT COUNT(*) FROM catcher_challenge_results) * 100 as overturned_pct
FROM catcher_challenge_results;
"""



splits_df = pd.read_sql_query(splits_query, conn)
splits_df.head()
```

LEADERS IN TOTAL CHALLENGE VALUE (CATCHERS)

```python
catcher_challenge_value_leaders_query = """
SELECT
player_name,
SUM(challenge_value) as total_challenge_value,
COUNT(*) as total_challenges,
COUNT(CASE WHEN description = 'ball' THEN 1 END) as total_overturned
FROM catcher_challenge_results
WHERE challenge_value IS NOT NULL
GROUP BY player_name
ORDER BY total_challenge_value DESC;
"""

catcher_challenge_value_leaders_df = pd.read_sql_query(catcher_challenge_value_leaders_query, conn)
catcher_challenge_value_leaders_df.head(20)
```

LEADERS IN AVERAGE CHALLENGE VALUE PER CHALLENGE (CATCHERS)

```python
average_challenge_value_leaders_query = """
SELECT
    player_name,
    AVG(challenge_value) AS average_challenge_value
FROM catcher_challenge_results
WHERE challenge_value IS NOT NULL
GROUP BY catcher_id
HAVING COUNT(*) >= 5
ORDER BY average_challenge_value DESC;
"""

average_challenge_value_leaders_df = pd.read_sql_query(average_challenge_value_leaders_query, conn)
average_challenge_value_leaders_df.head(20)
```

```python
# Find the rank of Valenzuela, Brandon in the average_challenge_value_leaders_df
rank = average_challenge_value_leaders_df[average_challenge_value_leaders_df['player_name'] == 'Valenzuela, Brandon'].index[0] + 1
print(f"Valenzuela, Brandon's rank: {rank}")
```
